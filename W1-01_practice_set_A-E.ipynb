```python
o = input("Take input: ")          # Q2 with string
v = "aeiouAEIOU"
v_o = ""
count = 0
for i in o:
  if i in v:
    v_o += i
    count += 1
print(f"Vowels: {v_o}")
print(f"Reversed String: {o[::-1]}")
print(f"Number of Vowels: {count}")
```

```python
o = input("Take input: ")          # Q2 with list
v = "aeiouAEIOU"
v_o = []
count = 0
for i in o:
  if i in v:
    v_o.append(i)
    count += 1
print(f"Vowels: {v_o}")
print(f"Reversed string: {o[::-1]}")
print(f"Number of Vowels: {count}")
```

```python
# Q3 grader

# first while inside loop take number. if it remains within the limit set by if condition then loop breaks.
# if not, invalid number appears & loop continues
# if conditions are true then the numbers grades are being checked & printed.

while True:
  n = input("Enter a number: ")
  if n.isdigit():
    n = int(n)
    if 0<=n<=100:
      break
  print("Invalid Input")

if n>=80:
  print("A Grade")
elif n>=70:
  print("B Grade")
elif n>=60:
  print("C Grade")
elif n>=50:
  print("D Grade")
else:
  print("F Grade")
```

```python
# total sum calculation from the list components. 1st say sum = 0, then with for loop price's each components i is added
# with the sum until loop ends with all the components added. then print happens

prices = [250, 80, 120, 45]
total_sum = 0
for i in prices:
  total_sum += i
print(f"total sum: {total_sum}")          # got stuck here for some time (print was inside loop)
```

```python
# here max calculation. each component of prices, is compared against biggest one by one.
# if the condition mets biggest value changes, if not loop continues till last. then print happens

prices = [250, 80, 120, 45]
biggest = prices[0]                 # got stuck here for some time
for i in prices:
  if i>biggest:
    biggest = i
print(f"biggest: {biggest}")   #q4/a+b
```

```python
# here min calculation. each component of prices, is compared against smallest one by one.
# if the condition mets smallest value changes, if not loop continues till last. then print happens

prices = [250, 80, 120, 45]
smallest = prices[0]    # got stuck here for some time
for i in prices:
  if i<smallest:
    smallest = i
print(f"smallest: {smallest}")   #q4/a+b
```

```python
## 1st a declared dictionary, then by looping for each component of the list for dictionary,
## it's count also declared. then printed & highest one also gotten

drink = ["tea", "coffee", "tea", "milk", "tea", "coffee"]
dict_drink = {}
for i in drink:
  if i in dict_drink:
    dict_drink[i] += 1
  else:
    dict_drink[i] = 1
print(dict_drink)

best = None
best_n = 0
for w in dict_drink:
    if dict_drink[w] > best_n:
        best = w
        best_n = dict_drink[w]
print("most frequent", best, best_n)   #q5/c
```

```python
## 1st i declared dictionary needed, then by looping for each component of the letter list for dictionary,
## it's numbers also declared. then printed & highest one also gotten using looping


letters = ["a", "b", "a", "c", "b", "a", "d"]

dict_letters = {}
for i in letters:
  if i in dict_letters:
    dict_letters[i] += 1
  else:
    dict_letters[i] = 1
print(dict_letters)

m_freq = None
m = 0

for n in dict_letters:
  if dict_letters[n] > m:
    m_freq = n
    m = dict_letters[n]
print(m_freq, m)                  #q5/c alter
```

```python
items = [("pen", 10), ("book", 120), ("bag", 450), ("cap", 90)]
t_qty = 0                     # declare total qty = 0
for name, qty in items:       # go through each items
  t_qty += qty                # total qty calc
avg = t_qty/len(items)        # avg calc
print(avg)
for name, qty in items:       # again loopin
  if qty > avg:               # > avg condition
    print(name)               # printing names above avg
# q6/D
```
