```python
# Here's your problem. It's a new one, so nothing in it is a copy of your calculator.

# Bill splitter

# A group eats out. Write one program that asks for the bill amount and the number of people, adds a 10% service charge, and works out what each person pays.

# Output. Print these five lines, clearly labelled, with money to 2 decimal places:

# Bill
# Service charge (10%)
# Total
# People
# Each person pays

# Edge case. If the number of people is 0, print a friendly message instead of the results. The program must not crash.

# Rules

# Write a numbered # plan first, then the code under it. Don't start coding until the plan is written.
# No AI and no Colab lesson open. Pick the conversion for each input yourself.
# Keep the whole program in one cell, not three copies.
# If you're stuck for 15 minutes, write # stuck: <why> and move on.

# Done when: these three tests all pass on that same program, one run each.

# Bill	People	Expected
# 1200	4	service 120.00, total 1320.00, each 330.00
# 999.5	3	service 99.95, total 1099.45, each 366.48
# 500	0	friendly message, no results, no crash

# Bring back: the plan, the code and the three outputs. Paste them here, or tell me to check the notebook as before. I'll review it, and I'll give hints only if you ask.
```

```python
# Plan
# 1. Ask for the bill amount and convert it to a number that allows decimals
# 2. Ask for the number of people and convert it to a whole number
# 3. If people is 0, print a friendly message
# 4. Otherwise, service charge = 10% of the bill
# 5. Total = bill + service charge
# 6. Each person pays = total divided by people
# 7. Print the five labelled lines with 2 decimal places (people is a whole number, so no decimals)


bill = float(input("Bill amount: "))
people = int(input("Number of people: "))      # i looked up took help ... since whole practiced class 3 of ostad, feeling somewhat exhausted & lethargic

if people == 0:
    print("people must be higher than zero")
else:
    service = bill * 0.1
    total = bill + service
    each = total / people

    print(f"Bill: {bill:.2f}")
    print(f"Service charge (10%): {service:.2f}")
    print(f"Total: {total:.2f}")
    print(f"People: {people}")
    print(f"Each person pays: {each:.2f}")
```
