```python
#1. Vowel count (easy: in, counter)
#Ask for a word. Print how many vowels it has (a, e, i, o, u, upper or lower case).
#Tests: Education → 5, rhythm → 0, AEIOU → 5.

word = input("Enter a word: ").strip().lower()  # taking word
vowels = "aeiou"           #vowels declared
word_n = ""   # string where vowels gather from word
count = 0     # how many vowels are there
i = 0                   # loop counter........ same as above
while i < len(word):    # loop head
  if word[i] in vowels:  # check condition ..... looked up
    word_n += word[i]    # update word_n .... looked up
    count += 1
  i+= 1
print(count)
print("vowels are", word_n)
```

```python
# 4. Word counter (medium to hard: loop + accumulator + order of checks)
# Keep asking for a word until the user types stop (any case). Ignore empty input: print empty and ask again, without counting it. At the end, print how many words were entered (not counting stop or empties) and how many of them start with a vowel.
# Tests, in this order: apple, Egg, banana, an empty line, STOP → 3 words, 2 start with a vowel.

vowels = "aeiou"
total = 0
vowel_words = 0

while True:
    word = input("Enter a word: ").strip().lower()       # this cod was over my head.... understood nothing.

    if word == "stop":
        break
    elif word == "":
        print("empty")
        continue

    total += 1
    if word[0] in vowels:
        vowel_words += 1

print("words:", total)
print("start with a vowel:", vowel_words)
```

```python
# 5. Palindrome machine (hard: all of it together)
# Keep asking for a phrase until the user types quit. For each phrase, ignore spaces and letter case, and print whether it reads the same backwards. When the user quits, print how many phrases were checked and how many were palindromes.
# Tests: Never odd or even → palindrome · Hello World → not a palindrome · A man a plan a canal Panama → palindrome, then quit → 3 checked, 2 palindromes.

checked = 0
palindromes = 0
                                            # this one same too. understood nothing
while True:
    phrase = input("Enter a phrase: ").strip().lower()

    if phrase == "quit":
        break

    cleaned = phrase.replace(" ", "")

    checked += 1
    if cleaned == cleaned[::-1]:
        print("palindrome")
        palindromes += 1
    else:
        print("not a palindrome")

print("checked:", checked)
print("palindromes:", palindromes)
```

```python
# Problem 1: Same ends (indexing + if)
# Ask for a word (assume it has at least 1 letter). Print same ends if its first and last letters are the same, ignoring case. Otherwise print different ends.
# Tests: Level → same ends · Python → different ends · a → same ends.

word = input("Word: ").strip().lower()
if word[0] == word[-1]:
    print("same ends")
else:
    print("different ends")

#   .lower() makes L and l equal.
# word[0] is the first letter, and word[-1] is the last letter.
# Level → level: l and l match, so it prints same ends. a has one letter, so first and last are the same letter.
```

```python
# Problem 2: Catch a vowel (in + while True + break)
# Keep asking for a single letter until the user types a vowel (a, e, i, o, u, either case). Then print Got followed by the letter in lowercase.
# Tests, in this order: x · an empty line · 7 · E → Got e.

while True:
    letter = input("A letter: ").strip().lower()
    if len(letter) == 1 and letter in "aeiou":
        print("Got", letter)
        break
    print("Try again")


#     while True asks forever. The break is the only way out.
# Valid input hits break, so print("Try again") below it is skipped. This is the same idea as your vowel loop today.
# The len(letter) == 1 check matters. "" in "aeiou" is True in Python, because an empty string is "inside" every string. Without the check, an empty line would count as a vowel. That's why your empty-line test is in the list.
# Trace: x → fails, Try again. Empty line → fails, Try again. 7 → fails, Try again. E → e, length 1 and in aeiou, prints Got e.
```

```python
# Problem 3: Reach 20 (accumulator + while condition)
# Keep asking for whole numbers and adding them up until the total is more than 20. Then print how many numbers were entered and the total.
# Test: 5, 8, 9 → 3 numbers, total 22.

total = 0
count = 0
while total <= 20:
    n = int(input("Number: "))
    total += n
    count += 1
print(count, "numbers, total", total)


# Two variables exist before the loop. total starts at 0 and goes up by the number. count starts at 0 and goes up by 1.
# The condition total <= 20 is checked before each pass. That's the answer to your open question from earlier.
# Trace: start total = 0. 5 → total 5. 8 → total 13. 9 → total 22. Now 22 <= 20 is False, so the loop ends. It prints 3 numbers, total 22.
```
