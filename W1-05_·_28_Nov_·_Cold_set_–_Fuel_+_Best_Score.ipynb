Problem 1: Fuel cost (two inputs, a guard, labelled output)

Ask for the trip distance in km, and for the car's mileage in km per litre. Both can have decimals.
The fuel price is fixed at 120 per litre. Keep it in a variable.
Work out the litres needed and the total cost.
Guard: if the mileage isn't a sensible value, print a clear message and skip the calculation. The message and your check must say the same thing.
Print three labelled lines: distance, litres needed and total cost. Show money and litres with 2 decimals.
Tests, as (distance, mileage): (150, 12), (60, 8), (100, 0), (100, -5). Work out the expected answers by hand first.

```python
# (i tried to do by myself entirely, but looked up for being stuck. Don't why, despite doing similar problems
# why always forget at the time of always doing something slightly modified or similar problem.
# this looping part with for loop i completely for got. now with help from AI this is it

# Plan
# 1. Keep the fuel price in a variable (120 per litre)
# 2. Ask for distance and mileage, both as floats
# 3. Guard: if mileage <= 0, print a message and skip the calculation
# 4. Otherwise litres = distance / mileage, cost = litres * price
# 5. Print three labelled lines with 2 decimals

fuel_price_per_litre = 120

tests = [(150, 12), (60, 8), (100, 0), (100, -5)]

for distance_km, mileage_km_per_litre in tests:
    print(f"Test: distance={distance_km}, mileage={mileage_km_per_litre}")

    if mileage_km_per_litre <= 0:
        print("Mileage must be greater than 0.")
    else:
        litres_needed = distance_km / mileage_km_per_litre
        total_cost = litres_needed * fuel_price_per_litre

        print(f"Distance: {distance_km:.2f} km")
        print(f"Litres needed: {litres_needed:.2f}")
        print(f"Total cost: {total_cost:.2f}")

    print()
```

Problem 2: Best score (the cart in a new shape)

Loop: ask for a subject name, and stop when the name is stop (ignore case and extra spaces). Otherwise ask for that subject's score, as a number.
At the end, print the highest score and which subject it belongs to. If two subjects tie, the first one entered wins.
Guard: if stop is typed first, print a friendly message and run no calculation.
Tests: Math 72, Physics 88, English 65, then stop; and stop typed first.

```python
# for problem 2 it's time out. so yes couldn't do
# but for sure faced similar problem like above one. now just for recording, with AI i'm pasting it here.

```

```python
# Plan
# 1. Before the loop: best_subject = None, best_score = None
# 2. Loop forever: ask for the subject name and strip outer spaces
# 3. If the name in lowercase is "stop", break
# 4. Otherwise ask for that subject's score as a float
# 5. If it is the first score, or strictly greater than best_score, update both variables
# 6. After the loop: if best_subject is still None, print a friendly message
# 7. Otherwise print the highest score and its subject


best_subject = None
best_score = None

while True:
    name = input("Enter subject name (stop to finish): ").strip().lower()
    if name == "stop":
        break

    score = float(input(f"Enter {name} score: "))

    if best_score is None or score > best_score:
        best_subject = name
        best_score = score

if best_subject is None:
    print("No subjects were entered, so there is no highest score.")
else:
    print(f"Highest score: {best_score:.2f}")
    print(f"Subject: {best_subject}")
```
