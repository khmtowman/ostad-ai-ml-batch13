```python
# Part 1: shopping cart, rebuilt from scratch

# Write it in a blank cell with no look-ups and no old notebook open. Put your numbered steps as # comments first, then the code. The program should:

# Ask for item names until the user types stop.
# Ask for the price of each item.
# Print the total, the number of items and the average price.
# If stop is typed first, print a friendly message and not crash.

# Run both tests in the same program:

# Items pen, book, bag, with prices 12, 30.5 and 57.5. The total should be 100.0, the count 3 and the average about 33.33.
# stop typed first. You should see your message and no error.
```

```python
# Plan

# ask item name & put them in list until stop is input from the user.
items = []
while True:
  n = input("Enter Items Names: ").strip().lower()
  if n == "stop":
    break
  items.append(n)
print(items)


# price ask & put them in list
price = []
for p in items:
  p = float(input(f"Enter {p} Price: "))   # here {p} forgot, stuck for more time, so looked up
  price.append(p)
print(price)

# calculation &  # if a stop typed first put a friendly message.
total = 0
for q in price:
  total +=q
if len(items) == 0:
  print("No items entered yet")
else:          # average defined when stop not input first..... stuck long time finally understood reason with hints AI help
  average = total/len(items)
  print()
  print("="*45)
  print()
  print(f"Total Price: {total:.2f}")
  print(f"Number of Items: {len(items)}")
  print(f"Average Price: {average:.2f}")
```
